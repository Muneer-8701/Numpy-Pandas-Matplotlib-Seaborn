```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
%matplotlib inline
```

# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* barplot
* countplot
* boxplot
* violinplot
* stripplot
* swarmplot




```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
tips.describe()
```

```python
tips.tail()
```

### Bar Plot

```python


sns.barplot(x='sex',y='total_bill',data=tips,hue='smoker',saturation=0.9,errcolor="g",errwidth=6) 
#data category ke according karna hue parameter use karo
plt.figure(figsize=(12,3))
sns.set_style("darkgrid") 
plt.show()


```

### Count Plot
countplots plots the count of the number of records by category

```python
tips = sns.load_dataset('tips')

sns.countplot(x = "sex",data=tips,hue="smoker",)
plt.show()
```

### Box Plot

```python
sns.boxplot(x="day",y="total_bill",data=tips,hue="sex",order=["Fri","Sat",
            "Sun","Thu"],showmeans=True,palette = 'coolwarm')
# mean show in green color
# Can do entire dataframe with (orient='h') then horizontal box

plt.show()
```

### Violin Plot

```python
sns.violinplot(x="day", y="total_bill", data=tips,hue='sex',palette='Set1')
```

### Strip Plot

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True)
```

### Swarm Plot

```python
sns.swarmplot(x="day", y="total_bill",hue='sex',data=tips, palette="Set1")
```

### Line Plot

```python
#limited data hoga tab graph aise bnega
var = [1,2,3,4,5,6,7]
var1 = [2,3,4,1,5,6,7]
x1 = pd.DataFrame({"var":var,"var1":var1})

sns.lineplot(x="var",y="var1",data=x1)

plt.show()
```

```python
var = pd.read_csv("penguins.csv")
var
#csv files ka 10rows ka data chahiye then var.head()





sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=var,hue='sex',
             style="sex",markers=["o",'*'])
#data differentiate karna hue parameter use karo
#different color use karne ke liye palette parameter use kare
 
sns.set_style("dark")
#use above function to change background color


plt.show()
```

### Histogram Plot

```python
sns.displot(var["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240],
            kde=True,color="r")
            
              

plt.show()
```

### Scatter Plot

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=var,hue='sex',
             style="sex",markers=["o",'*'] )

plt.show()
```

## Heatmap

In order for a heatmap to work properly, your data should already be in a matrix form, the sns.heatmap function basically just colors it in for you. For example:


```python
# Matrix form for correlation data
tips.corr()
```

```python
sns.heatmap(tips.corr(),cmap='coolwarm',annot=True,linecolor='black',linewidths=1)
```

### clustermap

The clustermap uses hierarchal clustering to produce a clustered version of the heatmap. For example:

```python
# More options to get the information a little clearer like normalization
sns.clustermap(tips.corr(),cmap='coolwarm',standard_scale=1)
```

### Pair Plot

```python
sns.pairplot(var)
plt.show()
```

```python
sns.pairplot(var,vars=["total_bill","tip"],hue="sex")
#kuch columns ka graph bana to use #vars parameter
#only x-axis mai yeh two parameter # x_vars = ["total_bill","tips"]
plt.show()

```

### Violin Plot

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",)

plt.show()
```

### Cat Plot

```python
sns.catplot(x="tip",y="size",data=var,hue="sex",kind="bar")
#use kind parameter "box","count","violin","strip","boxen","point"

plt.show()
```

### Multiple Plot

```python
fg = sns.facerGid(var,col="day",hue="sex")

fg.map(plt.scatter,"total_bill","tip").add_legend()

plt.show()
```

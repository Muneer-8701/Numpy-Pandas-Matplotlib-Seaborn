### Create Array with Random numbers

```python
# rand() is used to generate a random value between 0 and 1
var = np.random.rand(2,4)
var
```

```python
# randn() is used to generate a random value close to zero 
var = np.random.randn(5)
var
```

```python
# to generate random number having  a given range
var = np.random.randint(1,100,5)
var
```

```python
var = np.array([1,2,3,4,5])
np.random.shuffle(var)
var
```

## Numpy Indexing and Slicing

```python
# one dimensional
var = np.array([9,8,7,6])
var[2]
```

```python
# two dimensional
var = np.array([[9,8,7],[4,5,6]])
var[1,2]
```

```python
# three dimensional
var = np.array([[[9,8],[4,5]]])
var[0,1,1]
```

```python
var = np.array([1,2,3,4,5,6,7])
var[1:5]
```

```python
var[1:]
```

```python
var[:5]
```

```python
var[1:5:2]
```

```python
var[::2]
```

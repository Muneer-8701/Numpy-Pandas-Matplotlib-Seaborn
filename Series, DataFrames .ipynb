## Series
(1D Datastructure in Pandas )

```python
import numpy as np
import pandas as pd
```

```python
x = [3,4,5,6,7,8]   #series mai used anything dic,tupples,list
S1 = pd.Series(x)  # (x,index=['a','b,'c','d']) #index change karne ke liye
print(S1)
```

```python
print(type(S1))  
print(S1[0])
```

```python
#series mai used anything dic,tupples,list
labels = ['a','b','c']
my_data = [10,20,30]
arr = np.array(my_data)
dic = {'a': 10,'b':20,'c':20}
```

```python
pd.Series(data = my_data)
```

```python
pd.Series(data = my_data,index = labels)  # (x,index=['a','b,'c','d']) #index change karne ke liye
```

```python
pd.Series(arr)
```

```python
pd.Series(dic)
```

## DataFrame
(2D Datastructure in Pandas0

```python
# Dataframe by using dictionary
dic = {"Name": ['python','c','c++','java'],"rank" : [1,4,3,2]}
var1 = pd.DataFrame(dic) #(dic,columns=["rank"]) #column get karna 
print(var1)
```

```python
print(var1["Name"][2]) 
```

```python
print(var1["Name"]) 
```

```python
# DataFrame Using Numpy
from numpy.random import randn
np.random.seed(101)
```

```python
# You can also use df = pd.DataFrame(randn(5,4),index='A B C D E'.split(),columns='W X Y Z'.split())
df = pd.DataFrame(randn(5,4),index= ['A','B','C','D', 'E'],columns=['W', 'X','Y','Z'])
df
```

### Selection and Indexing


```python
# Extract one column then become Series
df['W'] 
```

```python
type(df['W'])
```

```python
type(df)
```

```python
# Pass a list of column names
df[['W','Z']]
```

```python
# Creating a new column
df['new'] = df['W'] + df['Y']
df
```

```python
# Removing Column
df.drop('new',axis=1,inplace=True)
df
```

```python
# Remove rows
df.drop('E',axis=0,inplace=True)
df
```

```python
# Selecting Row
df.loc['A']
```

```python
# Select based on of position instead of label 
df.iloc[2]
```

```python
# Selecting subset of rows and columns
df.loc['B','Y']
```

```python
df.loc[['A','B'],['W','Y']]
```

### Conditional Selection

```python
df>0
```

```python
df[df>0]
```

```python
df[df['W']>0]
```

```python
df[df['W']>0]['Y']
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

### Bar Plot

```python
x=["java","c","c++","python"]
y=[40,30,20,60]

plt.xlabel("language",fontsize=10)   
plt.ylabel("popularity",fontsize=10) 
plt.title("BAR PLOT",fontsize=10)
c=['b','g','y','m']       

plt.bar(x,y,width=0.4,color=c,edgecolor='r',linewidth=5,linestyle=":",
        label=["java","c","c++","python"])

plt.legend(loc=2)   #under label likhene ke liye rank1 python
plt.show() 

#plt.savefig('imagename',dpi=2000)   #agar graph ko save karna
```

### overlap bar plot

```python
x=["python","c","c++","java"]
y=[50,30,20,40]
z=[30,12,5,25]
 
plt.bar(x,y,width=0.4,color="r")  #plot.barh() #hori
plt.bar(x,z,width=0.4,color="y")
plt.show()
```

### Scatter Plot

```python
x = [1,2,3,4,5]
y = [2,4,6,7,8]

plt.xlabel("X-axis",fontsize=12)
plt.ylabel("Y-axis",fontsize=12) 
plt.title("Scatter Plot",fontsize=12)
plt.text(2,4,"java")

colors=['b','g','y','m','g']
sizes=[400,200,500,200,300]

plt.scatter(x,y,c=colors,s=sizes,edgecolor="r",linewidth=2) #marker="*"
plt.show()
```

### multiple scatter plot


```python
x = [1,2,3,4,5]
y = [2,4,6,7,8]
z = [3,5,7,9,3]

plt.xlabel("X-axis",fontsize=12)
plt.ylabel("Y-axis",fontsize=12) 
plt.title("Scatter Plot",fontsize=12)
colors=['b','r','y','m','r']
sizes=[400,200,500,200,300]

plt.scatter(x,y,c=colors,s=sizes,edgecolor="r",linewidth=2)
plt.scatter(x,z,c="g",s=sizes,edgecolor="r",linewidth=2)
plt.show()
```

### Histogram Plot

```python
no = [10,20,30,21 ,38 ,100 ,28, 23,45,60,90,200,58 ]

plt.xlabel("language",fontsize=10)
plt.ylabel("popularity",fontsize=10) 
plt.title("Histogram  PLOT",fontsize=10)
l=[10,20,30,40,50] 

plt.axvline(35,color='y')
plt.grid()

plt.hist(no,color='r',edgecolor='g',bins=l,cumulative=1)     
#cumulative =-1 (stair)
#orientation="horizontal"
#histtype='step'

plt.show()
```

### Pie Plot

```python
x = [30,50,80,98]
y = ['c','c++','java','python']   
ex= [0.2,0.0,0.0,0.0 ]
c=['r','b','y','g']

plt.pie(x,labels=y,colors=c,explode=ex,autopct='%0.1f%%',radius=0.8,
       labeldistance=1.2,wedgeprops={'edgecolor':'r','linewidth':3})
#counterclock=False #pie chart ko clockwise karna

plt.title('Pie Plot')
plt.legend(loc=2)
plt.show()
```

### Multiple Pie Plot

```python
x = [30,50,80,98]
x1 = [40,30,30,80]
y = ['c','c++','java','python']  

c=['r','b','y','m']

plt.pie(x,labels=y,radius=1.2)
plt.pie(x1,radius=.8,colors=c) 
#plt.pie([1],colors="w") #remove above x1 plot to show ring plot

plt.show()
```

### Subplot

```python
x=[1,2,3,4]
y=[1,2,3,4]
plt.subplot(2,2,1)   #(nrows,ncols,position)
plt.plot(x,y,'g--') #g*-  #go

x1=[2,3,4,5]
plt.subplot(2,2,2)  
plt.scatter(x,x1,c='g')
## saveimage     plt.savefig('image.jpg')


plt.subplot(2,2,3)
plt.pie(x)

x2=['a','s','d','f']
y2=[10,20,30,40]
plt.subplot(2,2,4)
plt.bar(x2,y2)


plt.show()
```

### Stem Plot

```python
x = [1,2,3,4,5]
y = [2,4,6,5,6]

plt.stem(x,y,linefmt=':',markerfmt="g+")
# orientation="horizontal" #horziontal graph
# use_line_collection = False
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,2,4,6,5]

plt.step(x,y,color="r",marker='o',ms=5,mfc='g')
plt.grid()
plt.show
```

###  Box Plot

```python
x = [10,20,30,40,50,60,120]  #meadian =40  #means=47

plt.boxplot(x,widths=0.3,labels=["BOX"],patch_artist=True,showmeans=True,
           whiskerprops=dict(color='r'))

#notch=True means change box shape
#patch_artist means box color blue



plt.show()
```

### multiple box

```python
x = [10,20,30,40,50,60]  
x1 = [20,40,60,80,110]
y = [x,x1]

plt.boxplot(y,widths=0.3,labels=["python","java"],patch_artist=True,whiskerprops =dict(color='r'))

#notch=True means change box shape
#patch_artist means box color blue



plt.show()
```

### Stack/Area Plot

```python
x = [1,2,3,4,5]
area1 = [2,3,2,5,4]
area2 = [2,3,4,5,6]
area3 = [1,3,4,5,3]

l=["x","area1","area2","area3"]
c=['b','g','y','r']

plt.stackplot(x,area1,area2,area2,area3,labels=l,colors=c)

plt.legend(loc=2)
plt.show()
```

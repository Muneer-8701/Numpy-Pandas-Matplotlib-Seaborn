## Bivariate and Multivariate Analysis
ek se jada column par analysis then bivariate analysis
three column ya us se jada par analysis then multivariate


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

### 1. Scatterplot (Numerical - Numerical)

```python
sns.scatterplot(x='total_bill',y='tip',data=tips)   # bivariate analysis
  # OR
#sns.scatterplot(x='total_bill',y='tip',data=tips,hue = 'sex',style='smoker')   # multivariate variate analysis
```

### 2. Bar plot (Numerical - Categorical)

```python
sns.barplot(x='sex',y='tip',data=tips,)   # bivariate analysis

# sns.barplot(x='sex',y='tip',data=tips,hue = 'smoker',)   # multivariate variate analysis

```

### 3. Box plot (Numerical - Categorical)

```python
sns.boxplot(x='sex',y='tip',data=tips,)   # bivariate analysis

#sns.boxplot(x='sex',y='tip',data=tips,hue = 'smoker',)   # multivariate variate analysis
```

### 4. Dist plot (Numerical - Categorical)

```python
#sns.distplot(x=tips['tip'])   # bivariate analysis

# here hue is not available we make hue parameter using extract

sns.distplot(x=tips[tips['sex']=='Male']['tip'],hist=False)
sns.distplot(x=tips[tips['sex']=='Female']['tip'],hist=False)

```
